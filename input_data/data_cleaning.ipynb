# KEN Model — Time-Series Data Cleaning

Utility notebook that cleans the raw SAMA time-series CSV files (`ts_*_raw.csv`) into the
model's input format (`ts_*.csv`). Run the cells in order.

Note: re-running overwrites the cleaned `input_data/ts_*.csv` files; confirm the outputs
match the committed versions before use.

```python
%reload_ext autoreload
%autoreload 2
import pandas as pd
from model.utils import to_float_or_nan
```

```python
TIMESERIES_KEYS = ["energy", "government",
                   "investment", "monetary", "national"]


def clean_timeseries_csvs(path):
    """Clean up sama file data."""

    path1 = 'input_data/ts_' + path + '_raw.csv'
    delimiter = ';'
    df = pd.read_csv(path1, delimiter=delimiter).transpose()

    # Set the first row as column names
    df.columns = [col.strip() if type(
        col) is str else col for col in list(df.iloc[0])]
    df = df[1:]

    # Drop columns with NaN column names
    df = df.drop(columns=[col for col in df.columns if pd.isna(col)])

    # Convert all values to float, or NaN if it fails
    df = df.map(to_float_or_nan)

    # Save cleaned file
    path2 = 'input_data/ts_' + path + '.csv'
    df.to_csv(path2, index=True)
```

```python
for key in TIMESERIES_KEYS:
    clean_timeseries_csvs(key)
```
